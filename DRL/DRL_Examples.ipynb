## Deep Q-Network (DQN)


Used method: [Stable Baselines3 DQN](https://stable-baselines3.readthedocs.io/en/master/modules/dqn.html)

Learning script: [DQN_training.py](DQN_training.py)

Load and run the trained model.

```python
from gymnasium_display_recorder import VideoWrapper, show
from itertools import count
from stable_baselines3 import DQN

env_orig = gym.make('LunarLander-v3', render_mode="rgb_array")
env = VideoWrapper(env_orig, 'DQN', render_fps=30)

model = DQN.load("dqn_final_model")

attempts = 5
for _ in range(attempts):
    state, info = env.reset()
    G = 0.0  # calculate undiscounted return for this episode
    for t in count():
        action, _states = model.predict(state, deterministic=True)
        observation, reward, terminated, truncated, _ = env.step(action.item())
        state = observation
        G += reward

        if terminated or truncated:
            break
    
    print(f"Return: {G:.2f}")
    show(env)

env.close()
```

Calculate the average reward of 100 tries.

```python
from tqdm import tqdm

env = gym.make('LunarLander-v3', render_mode=None)

attempts = 100
G = 0.0
for _ in tqdm(range(attempts)):
    state, info = env.reset()
    for t in count():
        action, _states = model.predict(state, deterministic=True)
        observation, reward, terminated, truncated, _ = env.step(action.item())
        state = observation
        G += reward

        if terminated or truncated:
            break
    
print("Average Return: {:.2f}".format(G/attempts))

env.close()
```
