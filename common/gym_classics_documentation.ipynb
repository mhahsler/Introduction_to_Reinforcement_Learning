```python
from gym_classics.envs.abstract.base_env import BaseEnv

help(BaseEnv)
```

### Important Internal Methods

To create a gym-classic environment that behaves differently than the standard environment, you will need to overwrite some of these 
internal methods. 

```python
from gym_classics.envs.abstract.base_env import BaseEnv

help(BaseEnv.__init__)
help(BaseEnv._next_state)
help(BaseEnv._reward)
help(BaseEnv._done)
```

**Note:** the `*` in front of `*random_elements` means that the method accepts a arbitrary number of positional arguments. If there are no
random elements, then this is just ignored.
