`<Your explanation>`

In high-dimensional observation spaces only a small number of states are encountered frequently
if we follow a (near) optimal policy. Essentially, only states on the optimal path will be visited. 
It is typically fine to make small mistakes for states that are rarely encountered using our policy. This idea can be implemented by index hash tables.
