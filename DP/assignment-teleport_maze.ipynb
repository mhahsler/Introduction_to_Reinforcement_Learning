## Introduction

We are working on a maze problem.
* The start and goal states are marked as `S` and `G`. 
* `T` is a teleport button which teleport the agent to the square marked `t`. 
* This is an **undiscounted** problem. 
* The step cost is -1 till the agent reaches the goal which ends the episode and no more reward (here cost) is accumulated. 

**Note:** There are two main methods to force an agent to learn the shortest path to the goal:

* Option 1: Using a constant negative reward of -1 for each step and no discounting. This is what we are doing here.
* Option 2: Use only a final large reward for getting to the goal and 0 otherwise. Here, we use a discounting with a factor $<1$.
