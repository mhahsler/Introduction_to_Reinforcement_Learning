# RL Schedules

Schedules are used in RL mainly for reducing $\alpha$ to help convergence and to reduce $\epsilon$ for on-polict methods 
to converge to the optimal deterministic policy.

```python
import numpy as np
np.set_printoptions(precision=2)

SEED = 42
rng = np.random.default_rng(SEED)
```

```python
import gymnasium as gym
import gym_classics2

from gym_classics2.animation import gridworld_animate                               
from gym_classics2.algorithms.policy import random_policy
```

You can read more on the implemented schedules in the [`gym_classics2` man pages](https://michael.hahsler.net/gym-classics2/api/algorithms/#schedules).

```python
from gym_classics2.algorithms.schedules import ConstantSchedule, StepSchedule, LinearDecaySchedule, \
    ExponentialDecaySchedule, InverseDecaySchedule, CosineAnnealingSchedule, plot_schedule     
```

```python
plot_schedule(ConstantSchedule(0.2))
plot_schedule(StepSchedule(1, 0.0, 500))
plot_schedule(LinearDecaySchedule(1, 0.0, 500))
plot_schedule(ExponentialDecaySchedule(1, 0.0, 0.99))
plot_schedule(InverseDecaySchedule(1, 0.0))
plot_schedule(CosineAnnealingSchedule(1, 0.0, 500))
plot_schedule(CosineAnnealingSchedule(1, 0.0, 500, warm_restarts=True))
```
